```python
import numpy as np
from scipy import stats

data = [25000, 31000, 20000, 28000, 29500, 32000, 21000, 30500,
        27000, 28500, 30000, 23000, 31000, 26000, 29500]

sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
n = len(data)

confidence = 0.95
alpha = 1 - confidence
t_critical = stats.t.ppf(1 - alpha/2, df=n-1)
margin_of_error = t_critical * (sample_std / np.sqrt(n))
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

print(f"Sample Mean: ₹{sample_mean:.2f}")
print(f"Sample Standard Deviation: ₹{sample_std:.2f}")
print(f"95% Confidence Interval: ₹{confidence_interval[0]:.2f} to ₹{confidence_interval[1]:.2f}")


claimed_mean = 30000
if confidence_interval[0] <= claimed_mean <= confidence_interval[1]:
    print("Fail to reject the null hypothesis: The average cost is consistent with ₹30,000.")
else:
    print("Reject the null hypothesis: The average cost is significantly different from ₹30,000.")
```
